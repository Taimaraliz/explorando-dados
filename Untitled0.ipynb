```python
sales_df.withColumn("num_pieces_sold_double", sales_df.num_pieces_sold.cast("double")).show()
```

```python
sales_df = sales_df.withColumn("num_pieces_sold", sales_df.num_pieces_sold.cast("double"))
```

```python
sales_df.groupBy("seller_id").sum("num_pieces_sold").show()
```

```python
agrupado = sales_df.groupBy("seller_id").sum("num_pieces_sold")
```

```python
agrupado.show()
```

```python
sellers_df.show()
```

```python
agrupado.join(sellers_df, on = "seller_id").show()
```

```python
agrupado.join(sellers_df, on = "seller_id")\
.withColumnRenamed("sum(num_pieces_sold)", "total_pecas_vendidas")\
.withColumnRenamed("seller_name", "nome_vendedor")\
.show()
```

```python
agrupado.join(sellers_df, on = "seller_id")\
.withColumnRenamed("sum(num_pieces_sold)", "total_pecas_vendidas")\
.withColumnRenamed("seller_name", "nome_vendedor")\
.select("nome_vendedor", "total_pecas_vendidas")\
.show()
```

```python
agrupado.join(sellers_df, on = "seller_id")\
.withColumnRenamed("sum(num_pieces_sold)", "total_pecas_vendidas")\
.withColumnRenamed("seller_name", "nome_vendedor")\
.select("nome_vendedor", "total_pecas_vendidas")\
.write.parquet("vandas_agrupadas")
```
