# Visualização de dados

```python
import pandas as pd
```

```python
dados = pd.read_csv("/content/drive/MyDrive/analise de dados programaria/analise_dados.csv")
```

```python
dados.head()
```

```python
genero_counts = dados["GENERO"].value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
plt.bar(height = genero_counts.values, x = genero_counts.index)
plt.title("Quantidade de pessoas por gênero na área de dados ")
plt.xlabel("Gênero")
plt.xlabel("Contagem")
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure()
sns.countplot(data = dados, x = "GENERO", palette = "pastel")
plt.title("Quantidade de pessoas por gênero na área de dados ")
plt.xlabel("Gênero")
plt.xlabel("Contagem")
plt.show()
```

```python
salario_por_idade = dados.groupby("IDADE")["SALARIO"].mean()
```

```python
salario_por_idade
```

```python
plt.figure()
plt.plot(salario_por_idade.index, salario_por_idade.values, marker="o", linestyle="--")
plt.title("Salário médio por idade")
plt.xlabel("Idade")
plt.ylabel("Salário médio")
plt.grid("True")
plt.show()
```

```python
import plotly.express as px
```

```python
fig = px.line( salario_por_idade.reset_index(), x = "IDADE", y = "SALARIO", title = "Média de salário por idade", markers = True)
fig.show()
```

```python
plt.figure(figsize=(15,5))
plt.scatter(dados["IDADE"], dados["SALARIO"], alpha=0.5)
plt.title("Diapersão de salário po idade")
plt.xlabel("Idade")
plt.ylabel("Salário")
plt.grid("True")
#plt.show() ##não é necessario do show pra visualizar o gráfico
```
